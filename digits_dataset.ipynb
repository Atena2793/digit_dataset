```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB, CategoricalNB
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv(r"A:\practice_files\digits_dataset.csv")
df.info()
```

```python
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
x = df.drop('target', axis=1)
y = df.target
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.15)
```

```python
gnb = GaussianNB().fit(x_train, y_train)
pred1 = gnb.predict(x_test)
cnb = CategoricalNB(min_categories=17).fit(x_train, y_train)
pred2 = cnb.predict(x_test)
rf = RandomForestClassifier().fit(x_train, y_train)
pred3 = rf.predict(x_test)
```

```python
print(f"GaussianNB is : %{accuracy_score(y_test, pred1)*100 :2f}")
print(f"CategoricalNB is : %{accuracy_score(y_test, pred2)*100 :2f}")
print(f"RandomForestClassifier is : %{accuracy_score(y_test, pred3)*100 :2f}")
```
